 ## 2) Data cleaning

 As the real world data usually contains missing values, duplicate records, invalid entries and incorrect data types, therefore we need to clean those data inorder to improve model performance.

no missing values for any columns

Having cholestrol 0 or less than 0 is medically invalid. So checking if there is any and if present remove that line.

```python
print(X) #consists of cholesterol data (predictor)
print(Y) #consists of heart disease data (target)
```

## 4) Algorithm Selection

Using logistic regression as algorithm because

```python
# importing logistic regression

from sklearn.linear_model import LogisticRegression
```

Creating a model 
Creates an empty Logistic Regression model 

```python
model_single = LogisticRegression() #No training occurs yet
```

## 5) Loss Function Selection 

To see how good or bad the models guess is we use something called Binary Cross-Entropy or Log Loss for short.
The formula for this is:
L=−[ylog(y')+(1−y)log(1−y')]

This formula helps us understand how off the models guess is from the real answer. If the model gets it wrong it gets penalized.. If it gets it right it gets a reward. The model is trying to adjust its settings to make the loss as small as possible. The problem with using one feature, like cholesterol level is that the model cannot understand really complicated relationships. This means it might not be very good at guessing and the loss value might be higher. As a result the models predictions might not be very accurate. Logistic regression, with one feature is limited so it might not give us the results. 

## 6) Model Learning(Training)

Importing Train-Test Split

```python
from sklearn.model_selection import train_test_split
```

Split data so that it consists : train = 80% and test= 20%
This is done so that the model can provide generalized data rather than fixated answer which is obtained after training and testing in the same dataset. So inorder to avoid that we separate 20% of data to test the model which is trained from the remaining 80% of data.

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2, 
    random_state= 42 #Ensures reproducible results
)
```

Train Model:
    The model learns coefficient, intercept by minimizing log-loss.

```python
print(X_train.dtypes)
print(Y_train.dtype)
print(type(model_single))
```

```python
model_single.fit(X_train, Y_train)
```

model_single.fit(X_train, y_train) teaches the Logistic Regression model to learn the relationship between cholesterol and heart disease by finding the best coefficient and intercept that minimize prediction error on the training data.

Displaying parameters

Coefficient:

Shows the influence of cholesterol

Positive coefficient => Higher cholesterol -> Higher heart disease probability
Negative coefficient => Higher cholesterol -> Lower heart disease probability

```python
print(model_single.coef_)

print(model_single.intercept_)
```

## 7) Model Evaluation 

```python
#importing metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

Prediciting Test data

```python
Y_pred = model_single.predict(X_test)
```

Accuracy:
Measures overall correctness

```python
accuracy = accuracy_score(Y_test, Y_pred)

print(accuracy)
```

Precision: Measures how many predicted heart disease cases are actually heart disease

```python
precision = precision_score(Y_test, Y_pred)

print(precision)
```

Recall : Measures how many actual heart disease patients were detected.

Important in healthcare because missing a sick patient is dangerous.

```python
recall = recall_score(Y_test, Y_pred)

print(recall)
```

F1 Score= Balances precision and recall.

```python
f1 = f1_score(Y_test, Y_pred)

print(f1)
```

Confustion matrix shows:
|            | Predicted No | Predicted Yes |
| ---------- | ------------ | ------------- |
| Actual No  | TN           | FP            |
| Actual Yes | FN           | TP            |


```python
cm = confusion_matrix(Y_test, Y_pred)

print(cm)
```

Classification Report: 
Provides:

Precision
Recall
F1-score
Support

for each class.

```python
print(classification_report(Y_test, Y_pred))
```

# Task 2 

Unlike Task 1, where only Cholesterol was used, here we use all available features. This allows the model to learn relationships from multiple risk factors such as Age, Blood Pressure, Maximum Heart Rate, and ST Depression.

# 3) Feature Design

Creating features and labels

```python
x = df.drop("Heart Disease", axis=1)

y = df["Heart Disease"]
```

```python
print(x.shape)
print(y.shape)
```

The above output means 270 patients and 13 input features.

### Feature Scaling

Since we have multiple input features so any large value in any of the features will be dominant in the model which will affect it. So inorder to have them in generalized form we scale them in common range.

```python
#import standard scaler
from sklearn.preprocessing import StandardScaler
```

Standard Scaler transforms each feature where mean will be 0 and distribution will be 1. z= (x−μ)/σ​

```python
scaler = StandardScaler()

x_scaled = scaler.fit_transform(x)
```

# 4) Algorithm Selection

Importing logistic regression 

```python
from sklearn.linear_model import LogisticRegression
```

Creating model with multiple iterations so that it gives the optimization algorithm enough iterations to converge

```python
model_multi = LogisticRegression(max_iter=1000)
```

## 5) Loss Function Selection 


Logistic Regression uses Binary Cross-Entropy, also known as Log Loss, to measure prediction error. This loss function evaluates how closely the predicted probabilities match the actual class labels. Incorrect predictions with high confidence receive a larger penalty, encouraging the model to produce more accurate probability estimates.

L=−[ylog(y')+(1−y)log(1−y')]

## 6) Model learning(training)

Same as task 1

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x_scaled,
    y,
    test_size=0.2,
    random_state=42
)
```

Train model


```python
model_multi.fit(x_train, y_train)
```

The model learns 13 coefficient (1 for each feature) and 1 intercept

```python
# Display intercept
print("Intercept:")
print(model_multi.intercept_)
```

```python
#Display coefficients
coef_df = pd.DataFrame({
    "Feature": x.columns,
    "Coefficient": model_multi.coef_[0]
})

coef_df.sort_values(
    by="Coefficient",
    ascending=False
)
```

Here +ve coefficient => Feature increases heart disease risk
-ve coefficient => Feature decreases heart disease risk

## Model Evaluation

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
#Make predictions
y_pred_multi = model_multi.predict(x_test)
```

```python
#Accuracy
accuracy_multi = accuracy_score(
    y_test,
    y_pred_multi
)

print("Accuracy:", accuracy_multi)
```

```python
#Precision 
precision_multi = precision_score(
    y_test,
    y_pred_multi
)

print("Precision:", precision_multi)
```

```python
#Recall
recall_multi = recall_score(
    y_test,
    y_pred_multi
)

print("Recall:", recall_multi)
```

```python
#F1 Score
f1_multi = f1_score(
    y_test,
    y_pred_multi
)

print("F1 Score:", f1_multi)
```

```python
#Confusion Matrix
cm_multi = confusion_matrix(
    y_test,
    y_pred_multi
)

print(cm_multi)
```

```python
#Classification Report

print(
    classification_report(
        y_test,
        y_pred_multi
    )
)
```
